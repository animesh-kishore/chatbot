```python
from openai import OpenAI
```

```python
# Endpoint to access ollama APIs
OLLAMA_BASE_URL = 'http://localhost:11434/v1'
OLLAMA_API_KEY = 'ollama'
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key=OLLAMA_API_KEY)
```

```python
# Init conversation. Charlie starts, folowed by Alex and Bob, thereafter round-robin
conversation = 'Charlie: Hi, I would like to make a friend\n'

# Template for Alex prompt
alex_system = 'You are snarky chatbot. You challenge and mock every response. Your name is Alex. You are in conversation with Bob and Charlie. Always format your response with prefix Alex:'
alex_user = f'Provide response based on below older conversation.\n{conversation}'

# Template for Bob prompt
bob_system = 'You are chatbot. You challenge every response in a very humble way. Your name is Bob. You are in conversation with Alex and Charlie. Always format your response with prefix Bob:'
bob_user = f'Provide response based on below older conversation.\n{conversation}'

# Template for Charlie prompt
charlie_system = 'You are chatbot. You avoid confrontation and try to come up with common grounds in most humble way possible. Your name is Charlie. You are in conversation with Alex and Bob. Always format your response with prefix Charlie:'
charlie_user = f'Provide response based on below older conversation.\n{conversation}'
```

```python
# Functions to log response of Alex, Bob and Charlie
MODEL = 'deepseek-v3.1:671b-cloud' # Largest model in ollama

def call_alex():
    global conversation
    messages = [{'role': 'system', 'content': alex_system}, {'role': 'user', 'content': alex_user}]
    response = ollama.chat.completions.create(model=MODEL, messages=messages)
    
    # Display response on screen and log it to conversation variable
    print(response.choices[0].message.content)
    conversation += f'{response.choices[0].message.content}\n'

def call_bob():
    global conversation
    messages = [{'role': 'system', 'content': bob_system}, {'role': 'user', 'content': bob_user}]
    response = ollama.chat.completions.create(model=MODEL, messages=messages)
    
    # Display response on screen and log it to conversation variable
    print(response.choices[0].message.content)
    conversation += f'{response.choices[0].message.content}\n'

def call_charlie():
    global conversation
    messages = [{'role': 'system', 'content': charlie_system}, {'role': 'user', 'content': charlie_user}]
    response = ollama.chat.completions.create(model=MODEL, messages=messages)
    
    # Display response on screen and log it to conversation variable
    print(response.choices[0].message.content)
    conversation += f'{response.choices[0].message.content}\n'
```

```python
print(conversation) # Charlie opens conversation
# Hold 5 turns of conversation
for i in range(5):
    #print(f'\nhistory: {conversation}')
    call_alex()
    #print(f'\nhistory: {conversation}')
    call_bob()
    #print(f'\nhistory: {conversation}')
    call_charlie()
    print('\n') # Add newline between each round
```
